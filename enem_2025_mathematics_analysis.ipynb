### Objectives

This project aims to:

- explore Mathematics performance in ENEM 2025;
- investigate differences in Mathematics performance across geographical and school-related characteristics;
- apply NumPy techniques to a large real-world dataset;

### 4.2 Loading the Data into NumPy Arrays

The variables required for each stage of the analysis are loaded into NumPy arrays according to their data types. The Mathematics score (`NU_NOTA_MT`) is loaded first as the main quantitative variable of the analysis.

```python
# Load Mathematics scores into a NumPy array
math_scores = np.genfromtxt(
    results_path,
    delimiter=";",
    skip_header=1,
    usecols=25,
    encoding="latin-1"
)
```

```python
# Inspect the Mathematics scores array
print("Type:", type(math_scores))
print("Dimensions:", math_scores.ndim)
print("Shape:", math_scores.shape)
print("Size:", math_scores.size)
print("Data type:", math_scores.dtype)
```

```python
# Count missing Mathematics scores
missing_scores = np.isnan(math_scores).sum()

print("Missing scores:", missing_scores)
```

```python
# Remove missing Mathematics scores using boolean indexing

valid_scores_mask = ~np.isnan(math_scores)
valid_math_scores = math_scores[valid_scores_mask]

print("Valid scores:", valid_math_scores.size)
```

```python
# Calculate descriptive statistics for Mathematics scores
print(f"Minimum score: {valid_math_scores.min():.2f}")
print(f"Maximum score: {valid_math_scores.max():.2f}")
print(f"Mean score: {valid_math_scores.mean():.2f}")
print(f"Standard deviation: {valid_math_scores.std():.2f}")
print(f"Median score: {np.median(valid_math_scores):.2f}")
```

```python
# Investigate Mathematics scores equal to zero using boolean indexing

# Create a boolean mask
zero_scores_mask = valid_math_scores == 0

# Select the scores that satisfy the condition
zero_scores = valid_math_scores[zero_scores_mask]

# Count the number of zero scores
zero_scores_count = zero_scores.size

print("Scores equal to zero:", zero_scores_count)
```

**Observation:** Among the 3,260,336 valid Mathematics scores, 893 were equal to zero, representing approximately 0.027% of the valid scores.

```python
# Identify participants with the maximum Mathematics score using boolean indexing

# Get the maximum score
maximum_score = valid_math_scores.max()

# Create a boolean mask
maximum_score_mask = valid_math_scores == maximum_score

# Select the maximum scores
maximum_scores = valid_math_scores[maximum_score_mask]

# Count the number of maximum scores
maximum_score_count = maximum_scores.size

print(f"Maximum score: {maximum_score:.2f}")
print("Number of maximum scores:", maximum_score_count)
```

**Observation:** The highest Mathematics score in ENEM 2025 was 980.30, achieved by 88 participants.

```python
# Calculate the quartiles of Mathematics scores

quartiles = np.percentile(valid_math_scores, [25, 50, 75])

print(quartiles)
```

**Observation:** The middle 50% of valid Mathematics scores ranged from 416.5 to 606.8, with a median score of 500.0.

```python
# Get the total number of valid Mathematics scores
total = valid_math_scores.size

# Scores from 0 to 100
scores_0to100_mask = (valid_math_scores >= 0) & (valid_math_scores < 100)
number0to100 = (valid_math_scores[scores_0to100_mask]).size
percentage0to100 = (number0to100 / total) * 100
print(f"{percentage0to100:.2f}%")

# Scores from 100 to 200
scores_100to200_mask = (valid_math_scores >= 100) & (valid_math_scores < 200)
number100to200 = (valid_math_scores[scores_100to200_mask]).size
percentage100to200 = (number100to200 / total) * 100
print(f"{percentage100to200:.2f}%")
```

**Observation:** The first score range, from 0 to 100, contains 893 observations. Since this is exactly the same number of scores equal to zero, there are no positive scores below 100. Additionally, no scores were found in the 100 to 200 range. This raises a new question: **what is the lowest positive Mathematics score in the dataset?**

```python
# Find the lowest positive Mathematics score
scores_higher_than_0_mask = valid_math_scores > 0
print(valid_math_scores[scores_higher_than_0_mask].min())
```

**Finding:** The lowest positive Mathematics score in the dataset is 312.6. Therefore, no valid Mathematics scores were observed between 0 and 312.6.

```python
# Scores from 300 to 400
scores_300to400_mask = (valid_math_scores >= 300) & (valid_math_scores < 400)
number300to400 = (valid_math_scores[scores_300to400_mask]).size
percentage300to400 = (number300to400 / total) * 100
print(f"{percentage300to400:.2f}%")

# Scores from 400 to 500
scores_400to500_mask = (valid_math_scores >= 400) & (valid_math_scores < 500)
number400to500 = (valid_math_scores[scores_400to500_mask]).size
percentage400to500 = (number400to500 / total) * 100
print(f"{percentage400to500:.2f}%")

# Scores from 500 to 600
scores_500to600_mask = (valid_math_scores >= 500) & (valid_math_scores < 600)
number500to600 = (valid_math_scores[scores_500to600_mask]).size
percentage500to600 = (number500to600 / total) * 100
print(f"{percentage500to600:.2f}%")

# Scores from 600 to 700
scores_600to700_mask = (valid_math_scores >= 600) & (valid_math_scores < 700)
number600to700 = (valid_math_scores[scores_600to700_mask]).size
percentage600to700 = (number600to700 / total) * 100
print(f"{percentage600to700:.2f}%")

# Scores from 700 to 800
scores_700to800_mask = (valid_math_scores >= 700) & (valid_math_scores < 800)
number700to800 = (valid_math_scores[scores_700to800_mask]).size
percentage700to800 = (number700to800 / total) * 100
print(f"{percentage700to800:.2f}%")

# Scores from 800 to 900
scores_800to900_mask = (valid_math_scores >= 800) & (valid_math_scores < 900)
number800to900 = (valid_math_scores[scores_800to900_mask]).size
percentage800to900 = (number800to900 / total) * 100
print(f"{percentage800to900:.2f}%")

# Scores from 900 to the maximum score
scores_900to_max_mask = (valid_math_scores >= 900) & (valid_math_scores <= maximum_score)
number900to_max = valid_math_scores[scores_900to_max_mask].size
percentage900to_max = (number900to_max / total) * 100

print(f"{percentage900to_max:.2f}%")
```

### 4.4 Loading Geographical Data

To analyze Mathematics performance across different locations, the state where each participant took the exam (`SG_UF_PROVA`) is loaded into a NumPy array. The order of the observations must be preserved so that each state remains aligned with its corresponding Mathematics score.

```python
# Load the exam state (UF) into a NumPy array

test_uf = np.genfromtxt(
    results_path,
    delimiter=";",
    skip_header=1,
    usecols=13,
    dtype=str,
    encoding="latin-1"
)
```

```python
# Inspect the exam state array

print("Dimensions:", test_uf.ndim)
print("Shape:", test_uf.shape)
print("Size:", test_uf.size)
print("Data type:", test_uf.dtype)
```

```python
# Check whether the state and Mathematics score arrays have the same size

test_uf.size == math_scores.size
```

```python
# Keep only the states associated with valid Mathematics scores

valid_test_uf = test_uf[valid_scores_mask]

# Identify the unique states in the dataset

unique_ufs = np.unique(valid_test_uf)

print(unique_ufs)
print("Number of states:", unique_ufs.size)
```

```python
# Calculate the mean Mathematics score for each state

uf_means = []

for uf in unique_ufs:
    uf_mask = valid_test_uf == uf
    uf_scores = valid_math_scores[uf_mask]
    uf_mean = uf_scores.mean()

    uf_means.append(uf_mean)

# Convert the list of state means into a NumPy array

uf_means = np.array(uf_means)

# Get the indices that would sort the mean scores in ascending order

sorted_indices = np.argsort(uf_means)

# Sort states and mean scores using the same indices

sorted_ufs = unique_ufs[sorted_indices]
sorted_means = uf_means[sorted_indices]

# Display states ordered by mean Mathematics score

for i in range(sorted_ufs.size):
    print(f"{sorted_ufs[i]}: {sorted_means[i]:.2f}")
```

### 4.5 Mathematics Performance by Exam State

The mean Mathematics score was calculated for each Brazilian state according to the state where the exam was taken (`SG_UF_PROVA`). The state averages were then sorted in ascending order to facilitate comparison.

The lowest mean score was observed in **Amazonas (AM)**, with **464.61 points**, followed by **Amapá (AP)** with 466.41 and **Pará (PA)** with 467.07.

At the other end of the distribution, **São Paulo (SP)** had the highest mean Mathematics score, with **560.34 points**, followed by **Santa Catarina (SC)** with 553.49 and **Rio Grande do Sul (RS)** with 549.08.

The difference between the highest and lowest state averages was **95.73 points**, indicating substantial variation in mean Mathematics performance across the states where the exam was taken.

It is important to note that `SG_UF_PROVA` represents the state where the participant took the exam and does not necessarily represent their state of residence or schooling.

### 4.6 School Administrative Dependency

The school administrative dependency variable (`TP_DEPENDENCIA_ADM_ESC`) is loaded to investigate differences in Mathematics performance according to the administrative category of the school.

```python
# Load the school administrative dependency into a NumPy array

school_dependency = np.genfromtxt(
    results_path,
    delimiter=";",
    skip_header=1,
    usecols=7,
    encoding="latin-1"
)
```

```python
# Check whether the school dependency and Mathematics score arrays have the same size

school_dependency.size == math_scores.size
```

```python
# Keep only the school dependency values associated with valid Mathematics scores

valid_school_dependency = school_dependency[valid_scores_mask]

# Check whether the filtered arrays have the same size

print("Valid Mathematics scores:", valid_math_scores.size)
print("Corresponding school dependency values:", valid_school_dependency.size)
```

```python
# Identify the unique school dependency values

unique_school_dependencies = np.unique(valid_school_dependency)

print(unique_school_dependencies)
```

```python
# Create a mask for records with available school dependency information

school_info_mask = ~np.isnan(valid_school_dependency)

# Keep only records with available school dependency information

school_dependency_clean = valid_school_dependency[school_info_mask]
school_math_scores = valid_math_scores[school_info_mask]

print("School dependency records:", school_dependency_clean.size)
print("Corresponding Mathematics scores:", school_math_scores.size)
```

```python
# Identify the school dependency categories after removing missing values

clean_dependencies = np.unique(school_dependency_clean)

print(clean_dependencies)
```

After removing missing values, 1,292,603 observations remained with both a valid Mathematics score and available school administrative dependency information.

According to the ENEM 2025 Microdata Dictionary, the categories of `TP_DEPENDENCIA_ADM_ESC` are:

- **1:** Federal
- **2:** State
- **3:** Municipal
- **4:** Private

```python
# Calculate the mean Mathematics score for each school dependency category

dependency_means = []

for dependency in clean_dependencies:
    dependency_mask = school_dependency_clean == dependency
    dependency_scores = school_math_scores[dependency_mask]
    dependency_mean = dependency_scores.mean()

    dependency_means.append(dependency_mean)

# Convert the list of dependency means into a NumPy array

dependency_means = np.array(dependency_means)

# Define the school dependency category labels

dependency_labels = np.array([
    "Federal",
    "State",
    "Municipal",
    "Private"
])

# Get the indices that would sort the mean scores in ascending order

sorted_indices = np.argsort(dependency_means)

# Sort school dependency categories and mean scores using the same indices

sorted_dependency_labels = dependency_labels[sorted_indices]
sorted_dependency_means = dependency_means[sorted_indices]

# Display school dependency categories ordered by mean Mathematics score

for i in range(sorted_dependency_labels.size):
    print(f"{sorted_dependency_labels[i]}: {sorted_dependency_means[i]:.2f}")
```

**Observation:** Mathematics performance varies substantially across school administrative dependency categories.

**State schools** presented the lowest mean Mathematics score, with **480.63 points**, followed by **municipal schools**, with **532.26 points**. Higher averages were observed among **federal schools**, with **586.95 points**, and **private schools**, which presented the highest mean score, at **620.32 points**.

The difference between the highest and lowest averages was **139.69 points**, with private schools averaging substantially higher Mathematics scores than state schools.

These results describe differences observed in the ENEM 2025 data and should not be interpreted as evidence that school administrative dependency itself causes differences in Mathematics performance.

### 4.7 School Location

The school location variable (`TP_LOCALIZACAO_ESC`) is loaded to investigate differences in Mathematics performance according to the location of the school.

```python
# Load the school location into a NumPy array

school_location = np.genfromtxt(
    results_path,
    delimiter=";",
    skip_header=1,
    usecols=8,
    encoding="latin-1"
)
```

```python
# Check whether the school location and Mathematics score arrays have the same size

school_location.size == math_scores.size

# Keep only the school location values associated with valid Mathematics scores

valid_school_location = school_location[valid_scores_mask]

# Check whether the filtered arrays have the same size

print("Valid Mathematics scores:", valid_math_scores.size)
print("Corresponding school location values:", valid_school_location.size)

# Identify the unique school location values

unique_school_locations = np.unique(valid_school_location)

print(unique_school_locations)
```

```python
# Create a mask for records with available school location information

school_location_info_mask = ~np.isnan(valid_school_location)

# Keep only records with available school location information

school_location_clean = valid_school_location[school_location_info_mask]
school_location_math_scores = valid_math_scores[school_location_info_mask]

# Check whether the filtered arrays have the same size

print("School location records:", school_location_clean.size)
print("Corresponding Mathematics scores:", school_location_math_scores.size)

# Identify the school location categories after removing missing values

clean_school_locations = np.unique(school_location_clean)

print(clean_school_locations)
```

After removing missing values, 1,292,603 observations remained with both a valid Mathematics score and available school location information.

According to the ENEM 2025 Microdata Dictionary, the categories of `TP_LOCALIZACAO_ESC` are:

- **1:** Urban
- **2:** Rural

Both school location categories are represented in the filtered dataset.

```python
# Calculate the mean Mathematics score for each school location category

location_means = []

for location in clean_school_locations:
    location_mask = school_location_clean == location
    location_scores = school_location_math_scores[location_mask]
    location_mean = location_scores.mean()

    location_means.append(location_mean)

# Convert the list of location means into a NumPy array

location_means = np.array(location_means)

# Define the school location category labels

location_labels = np.array([
    "Urban",
    "Rural"
])

# Display the mean Mathematics score for each school location category

for i in range(location_labels.size):
    print(f"{location_labels[i]}: {location_means[i]:.2f}")
```

The mean Mathematics score was calculated for each school location category, considering only observations with both a valid Mathematics score and available school location information.

Urban schools presented a mean Mathematics score of **516.27 points**, while rural schools had a mean score of **465.96 points**.

The difference between the two categories was **50.31 points**, with observations associated with urban schools presenting a higher average Mathematics score.

These results indicate an association between school location and Mathematics performance in the analyzed data. However, the difference should not be interpreted as evidence that school location itself causes higher or lower performance, since other socioeconomic, educational, and regional factors may also be associated with the observed scores.


### 4.8 Mathematics Performance by Exam Municipality

The municipality where the exam was taken (`NO_MUNICIPIO_PROVA`) is analyzed to investigate differences in mean Mathematics performance across Brazilian municipalities.

The analysis aims to identify the **50 exam municipalities with the highest mean Mathematics scores** in the ENEM 2025.

```python
# Load the exam municipality into a NumPy array

test_municipality = np.genfromtxt(
    results_path,
    delimiter=";",
    skip_header=1,
    usecols=11,
    dtype=str,
    encoding="latin-1"
)
```

```python
# Check whether the municipality and Mathematics score arrays have the same size

test_municipality.size == math_scores.size

# Keep only the municipalities associated with valid Mathematics scores

valid_test_municipality = test_municipality[valid_scores_mask]

# Check whether the filtered arrays have the same size

print("Valid Mathematics scores:", valid_math_scores.size)
print("Corresponding municipalities:", valid_test_municipality.size)
```

```python
# Identify the unique exam municipalities

unique_municipalities = np.unique(valid_test_municipality)
print(f"Number of unique exam municipalities: {unique_municipalities.size}")

# Calculate the mean Mathematics score for each exam municipality

municipality_means = []

for municipality in unique_municipalities:
    municipality_mask = valid_test_municipality == municipality
    municipality_scores = valid_math_scores[municipality_mask]
    municipality_mean = municipality_scores.mean()

    municipality_means.append(municipality_mean)

# Convert the list of municipality means into a NumPy array
municipality_means = np.array(municipality_means)
```

```python
# Identify the state corresponding to each exam municipality

municipality_ufs = []

for municipality in unique_municipalities:
    municipality_mask = valid_test_municipality == municipality
    municipality_uf = valid_test_uf[municipality_mask][0]

    municipality_ufs.append(municipality_uf)

# Convert the list of states into a NumPy array
municipality_ufs = np.array(municipality_ufs)

# Get the indices that sort municipality means in ascending order

sorted_indices = np.argsort(municipality_means)

# Sort municipalities, states, and their mean scores

sorted_municipalities = unique_municipalities[sorted_indices]
sorted_municipality_ufs = municipality_ufs[sorted_indices]
sorted_municipality_means = municipality_means[sorted_indices]


# Select the 50 municipalities with the highest mean scores

top_50_municipalities = sorted_municipalities[-50:]
top_50_ufs = sorted_municipality_ufs[-50:]
top_50_means = sorted_municipality_means[-50:]
```

```python
# Display the top 50 municipalities in descending order

for i in range(49, -1, -1):
    rank = 50 - i

    print(
        f"{rank}. {top_50_municipalities[i]} - "
        f"{top_50_ufs[i]}: {top_50_means[i]:.2f}"
    )
```

**Observation:** Among the 1,768 exam municipalities represented in the valid Mathematics score data, **Veranópolis (RS)** presented the highest mean score, with **617.67 points**, followed by **Lajeado (RS)** with **615.14 points** and **Valinhos (SP)** with **615.01 points**.

The ranking refers to the municipality where the exam was taken and does not necessarily represent the participant's municipality of residence or schooling.
